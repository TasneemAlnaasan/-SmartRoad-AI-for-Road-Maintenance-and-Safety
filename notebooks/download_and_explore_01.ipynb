```python
import pandas as pd
import sys
sys.path.append("..")

from config import REGION, REGION_RAW_DIR

table_name = "crashes"
domain = REGION["socrata_domain"]
dataset_id = REGION["tables"][table_name]["dataset_id"]

#building the url
url = f"https://{domain}/resource/{dataset_id}.csv?$limit=100000"

df = pd.read_csv(url)

# save path: data/raw/sf/crashes.csv
path = REGION_RAW_DIR / f"{table_name}.csv"
df.to_csv(path, index = False)

print (f"saved {len(df)} rows to {path}")
```
