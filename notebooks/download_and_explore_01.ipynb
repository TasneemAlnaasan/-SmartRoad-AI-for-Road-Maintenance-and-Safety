```python
import pandas as pd
import sys
sys.path.append("..")

from config import REGION, REGION_RAW_DIR


domain = REGION["socrata_domain"]

for table_name, table_info in REGION["tables"].items():
    dataset_id = table_info["dataset_id"]
    path = REGION_RAW_DIR / f"{table_name}.csv"
    if path.exists():
        continue 

    # build the URL
    url = f"https://{domain}/resource/{dataset_id}.csv?$limit=100000"
     # download the table
    df = pd.read_csv(url)                  

    
     # save the table
    df.to_csv(path, index=False)          

    print(f"{table_name}: {len(df)} rows -> {path}")
```
